# Conteneurisation de l'API avec Docker

## Objectif

Après avoir développé et validé l'API REST, l'étape suivante consiste à la conteneuriser à l'aide de Docker.

La conteneurisation permet d'encapsuler l'application, le modèle entraîné et toutes les dépendances dans un environnement isolé et reproductible. Cette approche garantit que l'application s'exécutera de manière identique sur n'importe quelle machine, indépendamment de sa configuration.

Docker constitue une étape essentielle dans une démarche MLOps puisqu'il facilite le déploiement, la portabilité et la maintenance des modèles de Machine Learning.

# Création du fichier `requirements.txt`

## Objectif

Le fichier `requirements.txt` recense l'ensemble des bibliothèques Python nécessaires au fonctionnement de l'application.

Lors de la construction de l'image Docker, ce fichier est utilisé afin d'installer automatiquement toutes les dépendances requises. Cette approche garantit la reproductibilité de l'environnement d'exécution et facilite le déploiement de l'API sur différentes machines.

L'utilisation d'un fichier de dépendances constitue une bonne pratique dans les projets Machine Learning et MLOps, car elle permet de maintenir un environnement cohérent entre les phases de développement, de test et de production.

# Création du fichier `Dockerfile`

## Objectif

Le `Dockerfile` décrit les différentes étapes nécessaires à la construction de l'image Docker de l'application.

Il indique notamment :
- l'image de base à utiliser ;
- le répertoire de travail du conteneur ;
- les fichiers à copier ;
- l'installation des dépendances Python ;
- l'exposition du port utilisé par l'API ;
- la commande à exécuter au démarrage du conteneur.

Grâce au `Dockerfile`, l'application peut être reconstruite automatiquement dans un environnement identique, garantissant ainsi sa portabilité et sa reproductibilité.

# Création du fichier `.dockerignore`

## Objectif

Le fichier `.dockerignore` permet d'exclure certains fichiers et dossiers du contexte de construction Docker.

Son utilisation réduit la taille de l'image, accélère la construction du conteneur et évite de copier des éléments inutiles tels que les notebooks, les fichiers temporaires, les environnements virtuels ou les métadonnées Git.

Cette pratique contribue à produire une image Docker plus légère, plus rapide à construire et plus facile à maintenir.

# Construction de l'image Docker

## Objectif

Après avoir défini le `Dockerfile` et le fichier `requirements.txt`, l'étape suivante consiste à construire l'image Docker de l'application.

Cette image contient l'ensemble des éléments nécessaires à l'exécution de l'API, notamment le code source, les dépendances Python et le modèle de Machine Learning retenu.

Une fois construite, cette image peut être exécutée sur n'importe quelle machine disposant de Docker, sans nécessiter d'installation manuelle des bibliothèques Python.

# Construction de l'image Docker

## Objectif

Après avoir développé et testé l'API REST avec FastAPI, l'étape suivante consiste à construire une image Docker contenant l'ensemble de l'application.

Une image Docker représente une version complète et autonome de l'application. Elle regroupe le code source, le modèle de Machine Learning, les dépendances Python ainsi que l'environnement d'exécution dans un package unique.

Cette approche garantit que l'application pourra être exécutée de manière identique sur n'importe quelle machine disposant de Docker, sans nécessiter d'installation manuelle des bibliothèques.

## Construction de l'image

La commande suivante : docker build -t intrusion-api . demande à Docker de lire le fichier `Dockerfile` situé à la racine du projet et de construire une image nommée **intrusion-api**.

Au cours de cette étape, Docker :

- télécharge l'image de base Python ;
- copie les fichiers du projet dans le conteneur ;
- installe les dépendances définies dans `requirements.txt` ;
- intègre le modèle de Machine Learning sauvegardé ;
- prépare l'environnement nécessaire à l'exécution de l'API.


## Vérification de la création de l'image

Une fois la construction terminée, il est possible de vérifier que l'image a été créée avec succès en affichant la liste des images Docker présentes sur la machine.

Cette vérification permet de confirmer que l'application est prête à être exécutée dans un conteneur.

docker images

# Exécution du conteneur Docker

## Objectif

Après la création de l'image Docker, celle-ci est exécutée sous la forme d'un conteneur.

Le conteneur constitue une instance active de l'image Docker. Il permet de lancer l'API FastAPI dans un environnement totalement isolé tout en conservant exactement la même configuration que celle définie lors de la construction de l'image.

Le paramètre `-p 8000:8000` établit une correspondance entre le port 8000 de la machine locale et le port 8000 utilisé par l'application à l'intérieur du conteneur. Cette redirection permet d'accéder à l'API depuis un navigateur ou un client HTTP.

docker run -p 8000:8000 intrusion-api

## Validation du déploiement

Lorsque le conteneur démarre correctement, le serveur Uvicorn est lancé automatiquement et héberge l'API FastAPI.

L'interface de documentation interactive générée par Swagger devient alors accessible à l'adresse suivante :

http://localhost:8000/docs

Cette interface permet de tester les différentes routes de l'API sans avoir recours à un outil externe.

# Résultat obtenu

L'API de détection d'intrusions a été déployée avec succès dans un conteneur Docker.

Grâce à cette conteneurisation, l'application, le modèle de Machine Learning et toutes les dépendances sont regroupés dans une image unique, garantissant un déploiement reproductible et indépendant de la configuration de la machine hôte.

Cette étape constitue une pratique fondamentale du MLOps puisqu'elle facilite le déploiement, la portabilité, la maintenance et la mise en production des modèles de Machine Learning.

# Orchestration de l'application avec Docker Compose

## Objectif

Après avoir conteneurisé l'API avec Docker, l'étape suivante consiste à orchestrer son exécution à l'aide de Docker Compose.

Docker Compose permet de décrire l'ensemble des services composant une application dans un unique fichier de configuration (`docker-compose.yml`). Il simplifie ainsi le démarrage, l'arrêt et la gestion des conteneurs.

Bien que notre application ne comporte actuellement qu'un seul service (l'API FastAPI), cette approche constitue une bonne pratique MLOps et facilite l'ajout futur d'autres services.

## Configuration de Docker Compose

Le fichier `docker-compose.yml` décrit les services nécessaires au fonctionnement de l'application.

Dans notre projet, un unique service est défini : l'API FastAPI.

Docker Compose se charge automatiquement de construire l'image Docker, de créer le conteneur et de configurer les ports nécessaires.

## Déploiement de l'application

Une fois le fichier `docker-compose.yml` créé, Docker Compose permet de construire automatiquement l'image si nécessaire, puis de lancer le conteneur en une seule commande.

Cette approche remplace les commandes `docker build` puis `docker run` exécutées séparément.

docker compose up --build

## Vérification du déploiement

Après le lancement de Docker Compose, le serveur Uvicorn démarre automatiquement et héberge l'API FastAPI.

L'API est alors accessible à l'adresse suivante :

http://localhost:8000/docs

Cette interface permet de tester les différents endpoints et de vérifier le bon fonctionnement de l'application.
