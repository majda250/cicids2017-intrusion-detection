# Comparaison des modèles

## Objectif

Cette étape a pour objectif de comparer les performances des différents modèles développés au cours de cette étude.

Les modèles sont répartis en deux catégories :

- les modèles de Machine Learning ;
- les modèles de Deep Learning.

Dans un premier temps, le meilleur modèle de chaque catégorie est sélectionné. Dans un second temps, ces deux modèles sont comparés afin d'identifier celui qui sera retenu pour la phase de déploiement MLOps.

## Comparaison des modèles de Machine Learning

Cette section compare les performances des modèles de Machine Learning selon les métriques suivantes :

- Accuracy
- Precision
- Recall
- F1-score
- Temps d'entraînement
- Temps de prédiction

## Chargement des résultats

Les performances des différents modèles ont été enregistrées automatiquement après leur entraînement dans un fichier CSV. Cette étape consiste à charger ces résultats afin de réaliser leur comparaison.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
results = pd.read_csv("results/model_results.csv")
```

## Comparaison des modèles de Machine Learning

Cette section compare les performances des modèles de Machine Learning développés dans cette étude :

- Régression Logistique
- Decision Tree
- Random Forest

Les modèles sont évalués selon plusieurs critères :

- Accuracy
- Precision
- Recall
- F1-score
- Temps d'entraînement
- Temps de prédiction

```python
ml_models = results[
    results["Model"].isin([
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ])
]

ml_models
```

## Classement des modèles de Machine Learning

Les modèles sont classés selon leur **F1-score**, qui constitue une métrique particulièrement pertinente pour les problèmes de classification multiclasse et les jeux de données déséquilibrés.

```python
ml_models.sort_values(
    by="F1-score",
    ascending=False
)
```

## Comparaison graphique des modèles de Machine Learning

Le graphique suivant illustre les performances des différents modèles de Machine Learning en termes de F1-score.

```python
ml_ranking = ml_models.sort_values(
    by="F1-score",
    ascending=False
)

ml_ranking
```

## Comparaison graphique des modèles de Machine Learning

Le graphique suivant illustre les performances des différents modèles de Machine Learning en termes de F1-score.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(
    ml_models["Model"],
    ml_models["F1-score"]
)

plt.title("Comparaison des modèles de Machine Learning")

plt.ylabel("F1-score")

plt.grid(axis="y")

plt.show()
```

## Sélection du meilleur modèle de Machine Learning

Le modèle présentant le F1-score le plus élevé est retenu comme meilleur modèle de Machine Learning.

```python
best_ml = ml_models.loc[
    ml_models["F1-score"].idxmax()
]

best_ml
```

## Comparaison des modèles de Deep Learning

Cette section compare les architectures de Deep Learning développées au cours du projet :

- Multi-Layer Perceptron (MLP)
- Deep Neural Network (DNN)

Les mêmes indicateurs de performance sont utilisés afin de garantir une comparaison cohérente avec les modèles de Machine Learning.

```python
dl_models = results[
    results["Model"].isin([
        "MLP",
        "DNN"
    ])
]

dl_models
```

## Classement des modèles de Deep Learning

Les modèles de Deep Learning sont classés selon leur F1-score afin d'identifier l'architecture la plus performante.

```python
dl_ranking = dl_models.sort_values(
    by="F1-score",
    ascending=False
)

dl_ranking
```

## Comparaison graphique des modèles de Deep Learning

Le graphique suivant présente les performances des modèles de Deep Learning.

```python
plt.figure(figsize=(6,5))

plt.bar(
    dl_models["Model"],
    dl_models["F1-score"]
)

plt.title("Comparaison des modèles de Deep Learning")

plt.ylabel("F1-score")

plt.grid(axis="y")

plt.show()
```

## Sélection du meilleur modèle de Deep Learning

Le modèle présentant les meilleures performances est retenu pour la comparaison finale.

```python
best_dl = dl_models.loc[
    dl_models["F1-score"].idxmax()
]

best_dl
```

## Comparaison finale

Les deux meilleurs modèles issus des familles Machine Learning et Deep Learning sont comparés afin d'identifier le modèle qui sera retenu pour la phase de déploiement.

```python
finalists = pd.DataFrame([
    best_ml,
    best_dl
])

finalists
```

## Comparaison finale des meilleurs modèles

Le graphique suivant compare les deux meilleurs modèles obtenus précédemment selon leur F1-score.

```python
plt.figure(figsize=(6,5))

plt.bar(
    finalists["Model"],
    finalists["F1-score"]
)

plt.title("Comparaison finale")

plt.ylabel("F1-score")

plt.grid(axis="y")

plt.show()
```

## Sélection du modèle final

Le modèle présentant le meilleur compromis entre performances de classification et coût de calcul est retenu pour la suite du projet.

```python
best_model = finalists.loc[
    finalists["F1-score"].idxmax()
]

print("Modèle retenu :", best_model["Model"])
```

## Sauvegarde du modèle retenu

Après la comparaison des différents modèles, le modèle présentant les meilleures performances est automatiquement sélectionné.

Cette étape consiste à copier ce modèle dans un dossier dédié (`best_model`). Cette organisation permet aux étapes suivantes du projet (déploiement avec FastAPI, Docker et Kubernetes) d'utiliser directement le meilleur modèle sans intervention manuelle.

```python
import os
import shutil

os.makedirs("best_model", exist_ok=True)
```

## Copie du modèle sélectionné

Selon le type du modèle retenu (Machine Learning ou Deep Learning), le fichier correspondant est automatiquement copié dans le dossier `best_model`.

```python
model_name = best_model["Model"]

print(f"Modèle sélectionné : {model_name}")
```

```python
if model_name in [
    "Logistic Regression",
    "Decision Tree",
    "Random Forest"
]:

    source = f"models/{model_name}.pkl"
    destination = "best_model/best_model.pkl"

    shutil.copy(source, destination)

    print("Le meilleur modèle Machine Learning a été sauvegardé.")
```
