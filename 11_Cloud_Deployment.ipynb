# Déploiement de l'API sur Render

## Objectif

Après avoir développé, testé et conteneurisé l'API de détection d'intrusions réseau, la dernière étape consiste à la déployer sur une plateforme cloud.

Le déploiement permet de rendre l'application accessible via Internet à l'aide d'une URL publique, sans nécessiter une exécution locale.

Dans ce projet, le déploiement est réalisé avec **Render**, une plateforme cloud qui permet d'héberger facilement des applications Docker.

Cette étape représente l'aboutissement du pipeline MLOps en mettant le modèle de Machine Learning à disposition des utilisateurs sous forme d'un service web.

## Publication du projet sur GitHub

Render récupère automatiquement le code source depuis un dépôt GitHub.

Avant le déploiement, l'ensemble des modifications du projet est envoyé vers le dépôt distant afin que la plateforme puisse construire l'image Docker la plus récente.

## Déploiement de l'application

Après la publication du projet sur GitHub, Render construit automatiquement une image Docker à partir du `Dockerfile`, puis crée un conteneur exécutant l'API.

Une fois le déploiement terminé, la plateforme fournit une URL publique permettant d'accéder à l'application depuis n'importe quel navigateur ou client HTTP.

L'URL ressemblera à :

https://cicids2017-intrusion-detection.onrender.com

Swagger sera accessible via :

https://cicids2017-intrusion-detection.onrender.com/docs

# Résultat

L'API de détection d'intrusions a été déployée avec succès sur la plateforme cloud Render.

Grâce à ce déploiement, le modèle de Machine Learning est désormais accessible via une API REST publique, sans nécessiter une installation locale de l'environnement d'exécution.

Cette étape finalise le pipeline MLOps du projet en assurant la mise en production du modèle dans un environnement cloud.
