# Intégration Continue et Déploiement Continu avec Jenkins

## Objectif

Après le développement, la conteneurisation et le déploiement de l'API sur une plateforme cloud, la dernière étape du pipeline MLOps consiste à automatiser les différentes tâches de développement.

Cette automatisation est réalisée à l'aide de Jenkins, un serveur d'intégration continue largement utilisé dans les environnements DevOps et MLOps.

L'objectif est de construire une pipeline capable d'exécuter automatiquement les principales étapes du projet à chaque modification du code source.

# Pourquoi Jenkins ?

Jenkins permet d'automatiser les différentes étapes du cycle de développement logiciel.

Dans le cadre de ce projet, Jenkins présente plusieurs avantages :

- automatisation des tâches répétitives ;
- réduction des erreurs humaines ;
- exécution systématique des tests ;
- construction automatique de l'image Docker ;
- préparation au déploiement continu.

L'utilisation de Jenkins améliore ainsi la fiabilité et la reproductibilité du pipeline MLOps.

# Installation de Jenkins avec Docker

## Objectif

Afin d'automatiser les différentes étapes du pipeline MLOps, Jenkins est installé sous la forme d'un conteneur Docker.

Cette approche présente plusieurs avantages :

- aucune installation locale de Jenkins n'est nécessaire ;
- l'environnement est entièrement reproductible ;
- Jenkins peut être facilement mis à jour ou supprimé ;
- l'installation reste cohérente avec l'architecture Docker utilisée pour le projet.

## Création d'un volume persistant

Jenkins stocke l'ensemble de ses configurations, utilisateurs, plugins et pipelines dans un espace de stockage dédié.

Afin de conserver ces informations même après l'arrêt ou la suppression du conteneur, un volume Docker persistant est créé.

## Personnalisation de l'environnement Jenkins

L'image officielle de Jenkins ne contient pas le client Docker, ce qui empêche l'exécution des commandes nécessaires à la construction des images de conteneurs.

Afin de permettre à Jenkins d'interagir avec Docker, une image personnalisée a été créée à partir de l'image officielle `jenkins/jenkins:lts`. Cette image intègre le client Docker tout en conservant les fonctionnalités de Jenkins.

Cette approche garantit un environnement reproductible et adapté à l'automatisation des tâches de construction et de validation du projet.
